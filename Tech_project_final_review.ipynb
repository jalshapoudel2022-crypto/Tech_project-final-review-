```python
import pandas as pd

# Enter the data for the 10 selected students
data = {
    "student_id": [107, 590, 706, 1056, 1502, 1601, 2414, 2469, 2587, 2654],
    "study_hours": [7.4, 7.15, 7.06, 8.23, 9.55, 7.55, 4.65, 4.55, 6.22, 9.34],
    "grade": [78.92, 62.4, 81.66, 64.62, 79.62, 66.06, 65.05, 59.68, 50.54, 74.24]
}

# Create a table from the data
df = pd.DataFrame(data)

# Display the dataset
print(df)

# Check the number of rows and columns
print("Dataset size:", df.shape)

# Check for missing values
print("\nMissing values:")
print(df.isnull().sum())

# View basic statistics
print("\nSummary of the data:")
print(df.describe())

import matplotlib.pyplot as plt

plt.scatter(df["study_hours"], df["grade"])

plt.xlabel("Study Hours")
plt.ylabel("Grade")
plt.title("Relationship Between Study Hours and Grades")

plt.show()
```

```python
# Calculate the correlation between study hours and grades
correlation = df["study_hours"].corr(df["grade"])

print("Correlation:", round(correlation, 2))

import matplotlib.pyplot as plt
import numpy as np

x = df["study_hours"]
y = df["grade"]

# Calculate the trend line
slope, intercept = np.polyfit(x, y, 1)

# Create the scatter plot
plt.scatter(x, y, label="Students")

# Draw the trend line
plt.plot(x, slope * x + intercept, label="Trend line")

plt.xlabel("Study Hours")
plt.ylabel("Grade")
plt.title("Study Hours vs Grades")
plt.legend()
plt.show()
```

```python
# Calculate the average study hours
average_hours = df["study_hours"].mean()

# Calculate the average grade
average_grade = df["grade"].mean()

print("Average study hours:", round(average_hours, 2))
print("Average grade:", round(average_grade, 2))
```

```python
# Calculate the median study hours
median_hours = df["study_hours"].median()

# Calculate the median grade
median_grade = df["grade"].median()

print("Median study hours:", median_hours)
print("Median grade:", median_grade)
```

```python
# Calculate the range of study hours
range_hours = df["study_hours"].max() - df["study_hours"].min()

# Calculate the range of grades
range_grade = df["grade"].max() - df["grade"].min()

print("Range of study hours:", round(range_hours, 2))
print("Range of grades:", round(range_grade, 2))
```

```python
# Calculate standard deviation
std_hours = df["study_hours"].std()
std_grade = df["grade"].std()

print("Standard deviation of study hours:", round(std_hours, 2))
print("Standard deviation of grades:", round(std_grade, 2))
```

```python
print("Number of students:", len(df))

print("Missing values:")
print(df.isnull().sum())

print("Duplicate student IDs:")
print(df["student_id"].duplicated().sum())
```
